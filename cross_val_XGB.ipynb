```python
from xgboost import XGBRegressor
import pandas as pd

import os

print(os.listdir('/'))

!find / -name train.csv 2>/dev/null

train = pd.read_csv("/content/train.csv")

print(train.shape)

# print(train.columns)

train["TotalArea"] = (
    train["GrLivArea"]
    + train["TotalBsmtSF"]
)

train["HouseAge"] = (
    2026 - train["YearBuilt"]
)

train["TotalBath"] = (
    train["FullBath"]
    + train["HalfBath"] * 0.5
)

features = [
    'OverallQual',
    'GrLivArea',
    'GarageCars',
    'TotalBath',
    'GarageArea',
    'TotalArea',
    'HouseAge',
    'LotArea'
]

features.append("OverallCond")

features = [
    'OverallQual',
    'GrLivArea',
    'GarageCars',
    'TotalBath',
    'GarageArea',
    'TotalArea',
    'HouseAge',
    'LotArea',
    'OverallCond',
    'TotRmsAbvGrd',
    'Fireplaces'
]

features = [
    'OverallQual',
    'GrLivArea',
    'GarageCars',
    'TotalBath',
    'GarageArea',
    'TotalArea',
    'HouseAge',
    'LotArea',
    'OverallCond',
    'TotRmsAbvGrd',
    'Fireplaces',
    'LotFrontage'     # 新增
]

missing = (
    train[features]
    .isnull()
    .sum()
    .sort_values(ascending=False)
)

print("Missing Values:")
print(missing)

X = train[features]
# y = train['SalePrice']

import numpy as np

y = train["SalePrice"]

from xgboost import XGBRegressor
from sklearn.model_selection import cross_val_score

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    random_state=42
)

param_grid = {
    "n_estimators": [50, 100, 200],
    "max_depth": [3, 5, 10]
}

grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error"
)

grid.fit(X, y)

print("Best Params:", grid.best_params_)
print("Best MAE:", -grid.best_score_)

best_model = grid.best_estimator_

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": best_model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(importance_df.to_string(index=False))

importance_df["Importance_%"] = (
    importance_df["Importance"] * 100
)
```
